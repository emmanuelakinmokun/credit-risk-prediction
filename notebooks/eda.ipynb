## Dataset Columns Description

Here is the breakdown of each feature and target variable in the "Give Me Some Credit" dataset:

* **`SeriousDlqin2yrs`**: The target label—whether the person experienced serious delinquency (90 days past due or worse) in the last 2 years ($1 = \text{Yes}, 0 = \text{No}$).
* **`RevolvingUtilizationOfUnsecuredLines`**: Total balance on credit cards and personal lines divided by the total credit limit (credit utilization ratio).
* **`age`**: Age of the borrower in years.
* **`NumberOfTime30-59DaysPastDueNotWorse`**: Number of times the borrower has been 30–59 days past due on a loan in the past 2 years.
* **`DebtRatio`**: Monthly debt payments, alimony, and living costs divided by monthly gross income.
* **`MonthlyIncome`**: The borrower's monthly income.
* **`NumberOfOpenCreditLinesAndLoans`**: Number of open loans (e.g., car loans, mortgages) and lines of credit (e.g., credit cards).
* **`NumberOfTimes90DaysLate`**: Number of times the borrower has been 90 days past due or worse on a payment.
* **`NumberRealEstateLoansOrLines`**: Number of mortgage and real estate loans (including home equity lines of credit).
* **`NumberOfTime60-89DaysPastDueNotWorse`**: Number of times the borrower has been 60–89 days past due in the past 2 years.
* **`NumberOfDependents`**: Number of family dependents living with the borrower (excluding themselves).
* **`Unnamed: 0`**: An accidental index column left over from saving the CSV file (recommended to drop).

## Data Cleaning & Anomaly Detection

* **Age Anomaly (`age`):** Found **1 record** with an age of `0`, which is a data entry error. This row was removed.
* **Credit Utilization Outliers (`RevolvingUtilizationOfUnsecuredLines`):** While utilization ratios can exceed 100% (1.0) due to over-limit fees, values scaled up to 50,708. These extreme values will be monitored as potential outliers.
* **Special Placeholder Codes (`98`):** The value `98` appeared repeatedly in the past-due history columns (`NumberOfTime30-59DaysPastDueNotWorse`, etc.). Investigation confirmed this is a dataset creator code for a special status or missing value rather than a literal count of 98 times past due. These were normalized to `0`.
* **Class Imbalance:** The target variable (`SeriousDlqin2yrs`) has a mean of ~6.7%, indicating a heavily imbalanced dataset where non-defaulters vastly outweigh defaulters.

```python
# Replace both 98 and 96 placeholder codes with 0 in past-due columns
past_due_cols = [
    'NumberOfTime30-59DaysPastDueNotWorse',
    'NumberOfTimes90DaysLate',
    'NumberOfTime60-89DaysPastDueNotWorse'
]

for col in past_due_cols:
    df[col] = df[col].replace([98, 96], 0)

df.describe()
```
